```python
import pandas as pd

df = pd.read_json("../data/threads_per_group.jsonl", lines=True, orient="records")
df["num_threads"] = df.thread_data.apply(len)
df["news_group"] = df.mbox_file.apply(lambda x: x[:-5])
del df["mbox_file"]
df
```

# Top 20 newsgroups 
The 20 news groups with the most threads 

```python
df.sort_values("num_threads", ascending=False).head(20)
```

# Top 20 longest threads

```python
threads_df = pd.DataFrame(
    [
        {"news_group": tup.news_group, **thread_data}
        for tup in df.itertuples()
        for thread_data in tup.thread_data
    ]
).sort_values("num_messages", ascending=False)
threads_df.head(20)
```

# Threads across newsgroups

```python
for subject, df_ in threads_df.head(200).groupby("thread_subject"):
    print(
        f"{subject}\n\tNews groups: {df_.news_group.to_list()}\n\tNum messages: {df_.num_messages.sum()}\n"
    )
```
