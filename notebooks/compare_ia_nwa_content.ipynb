```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib_venn import venn2
```

```python
df = pd.read_csv("../data/ia_nwa_content_comparison.csv").sort_values(
    "both", ascending=False
)
df.head(20)
```

```python
ia_only = df["ia_only"].sum()
nwa_only = df["nwa_only"].sum()
both = df["both"].sum()

print(f"IA only:  {ia_only:,}")
print(f"NWA only: {nwa_only:,}")
print(f"Both:     {both:,}")
```

```python
fig, ax = plt.subplots(figsize=(6, 6))
venn2(
    subsets=(ia_only, nwa_only, both),
    set_labels=("IA", "NWA"),
    ax=ax,
)
ax.set_title("Message overlap between IA and NWA (exact body match)")
plt.tight_layout()
plt.show()
```
