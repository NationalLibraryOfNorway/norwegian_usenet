```python
import pandas as pd
import matplotlib.pyplot as plt
from usenet_no.notebook_utils import venn2_fmt


def load_content(path):
    return pd.read_csv(path).sort_values("both", ascending=False)


full = load_content("../data/ia_nwa_content_comparison.csv")
filtered = load_content("../data/ia_nwa_content_comparison_date_filtered.csv")
```

# IA vs NWA Content Comparison

Exact-body-match message overlap between IA and NWA, for full IA data and IA filtered to the NWA date span.

```python
for label, df in [("Full IA", full), ("IA date-filtered", filtered)]:
    print(f"\n{label}")
    display(df.head(20))
```

```python
for label, df in [("Full IA", full), ("IA date-filtered", filtered)]:
    print(f"{label}:")
    print(f"  IA only:  {df['ia_only'].sum():,}")
    print(f"  NWA only: {df['nwa_only'].sum():,}")
    print(f"  Both:     {df['both'].sum():,}")
    print()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
for ax, (label, df) in zip(axes, [("Full IA", full), ("IA date-filtered", filtered)]):
    venn2_fmt(
        subsets=(df["nwa_only"].sum(), df["ia_only"].sum(), df["both"].sum()),
        set_labels=("NWA", "IA"),
        ax=ax,
        show_pct=True,
    )
    ax.set_title(f"Message overlap — {label}\n(exact body match)")
plt.tight_layout()
plt.show()
```

```python
sum_i = (filtered.nwa_only + filtered.ia_only).sort_values(ascending=False).index
sortby_sum = filtered.loc[sum_i]
sortby_sum = sortby_sum[(sortby_sum.nwa_only > 1) & (sortby_sum.ia_only > 1)]
top_50_sum = sortby_sum.head(50)
top_50_sum.to_json(
    "../data/newsgroups_for_selection.jsonl", index=False, lines=True, orient="records"
)
```
